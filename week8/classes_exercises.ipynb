```python
# “Create a class that represents a simple resource (like a Dataset loader) with attributes and methods.”
class Dataset:
    def __init__(self, data):
        self.data = data

    def count(self):
        return len(self.data)

    def mean(self):
        return sum(self.data) / self.count() if self.count() > 0 else None

# Usage
ds = Dataset([10, 20, 30])
print(ds.count())  # 3
print(ds.mean())   # 20.0
```

```python
# “Create a class hierarchy for model evaluators, e.g., base Evaluator and specific classes like AccuracyEvaluator.”
# • Interviewer wants to test inheritance, overriding, and polymorphism — core OOP concepts.

class Evaluator:
    def evaluate(self, y_true, y_pred):
        raise NotImplementedError("Must override")

class AccuracyEvaluator(Evaluator):
    def evaluate(self, y_true, y_pred):
        correct = sum(1 for a, b in zip(y_true, y_pred) if a == b)
        return correct / len(y_true)

# Usage
acc = AccuracyEvaluator()
print(acc.evaluate([1,0,1], [1,0,0]))  # 2/3 = 0.666…
```

```python
# “When do you use @classmethod or @staticmethod?”
# • @staticmethod: behavior that doesn’t access class or instance state
# • @classmethod: method that uses class (e.g., alternate constructors)

#Example: Alternate constructor
class ModelConfig:
    def __init__(self, learning_rate, batch_size):
        self.learning_rate = learning_rate
        self.batch_size = batch_size

    @classmethod
    def from_dict(cls, config):
        return cls(config["lr"], config["bs"])

config = {"lr": 0.001, "bs": 32}
mc = ModelConfig.from_dict(config)
print(mc.learning_rate)
```

```python
# Interviewers for a mid/AI role might ask you to define a base class that 
# can't be instantiated itself but must be subclassed — good for model pipelines or plug-in models.

#🧩 Example: Abstract Model Interface
from abc import ABC, abstractmethod

class BaseModel(ABC):
    @abstractmethod
    def train(self, X, y):
        pass

    @abstractmethod
    def predict(self, X):
        pass

class DummyModel(BaseModel):
    def train(self, X, y):
        self.mean = sum(y) / len(y)

    def predict(self, X):
        return [self.mean for _ in X]

# Usage
dm = DummyModel()
dm.train([1,2,3], [1,1,1])
print(dm.predict([0,0]))

#• @abstractmethod enforces that subclass must define train/predict
```

```python
# 🧠 5) Composition > Inheritance
# Real tasks sometimes require you to design classes that use other classes inside them — a key design pattern

# 🧩 Example: Training Pipeline
class Preprocessor:
    def fit(self, data):
        self.mean = sum(data)/len(data)

    def transform(self, data):
        return [x - self.mean for x in data]

class Pipeline:
    def __init__(self):
        self.preprocessing = Preprocessor()

    def fit_transform(self, data):
        self.preprocessing.fit(data)
        return self.preprocessing.transform(data)

# Usage
p = Pipeline()
print(p.fit_transform([10, 20, 30]))

# Why this matters:
# Shows you can organize responsibilities properly — interviewers really pay attention to clean separation of concerns.
```

```python
# 🧠 6) Magic/Dunder Methods
# Mid-level interviews may go deeper and ask about special methods like __str__, __repr__, __len__, etc.

#🧩 Example: Custom repr
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

p = Point(3, 5)
print(p)  # Point(3, 5)

# Be ready to implement comparisons like __eq__, or even __add__ to combine objects.
```

```python
# 🧠 7) Real-world Task: Build a Simple Logger Class
# “Implement a Logger class with levels (INFO, WARN, ERROR) and file output.”

# This tests:
# • Class design
# • Method logic
# • Optional state (log level)
# • Real functional output

# 🧩 Skeleton Implementation
class Logger:
    LEVELS = ("INFO", "WARN", "ERROR")

    def __init__(self, level="INFO"):
        self.level = level

    def log(self, msg, level="INFO"):
        if Logger.LEVELS.index(level) >= Logger.LEVELS.index(self.level):
            print(f"[{level}]: {msg}")

logger = Logger("WARN")
logger.log("This is a test", "INFO")   # ignored
logger.log("This is important", "ERROR")
```

```python
# 🧠 8) Linked Lists / Custom Data Structures
# For mid-level coding rounds, you might be asked to implement a small data structure with classes
#  — e.g., linked list nodes.

# 🧩 Example: Linked List Node
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, value):
        if not self.head:
            self.head = Node(value)
        else:
            cur = self.head
            while cur.next:
                cur = cur.next
            cur.next = Node(value)

```
